```python
# 初始化 dict

user_info = {"name": "Harry", "age": 20}
print(user_info)  # 輸出: {'name': 'Harry', 'age': 20}
```

```python
#資料較多時 建議寫法
fruit_price = {
    "apple": 100, 
    "banana": 50, 
    "orange": 80}
print(fruit_price)  # 輸出: {'apple': 100, 'banana': 50, 'orange': 80}
```

```python
# 印出 蘋果的價格
print(fruit_price["apple"])  # 輸出: 100
```

```python
# 修改 橘子的價格
fruit_price["orange"] = 90
print(fruit_price)  # 輸出: {'apple': 100, 'banana': 50, 'orange': 90}
```

```python
#新增 kiwi
fruit_price["kiwi"] = 150
print(fruit_price)  # 輸出: {'apple': 100, 'banana': 50, 'orange': 90, 'kiwi': 150}
```

```python
# 刪除 香蕉 dict.pop

fruit_price.pop("banana")
print(fruit_price)  # 輸出: {'apple': 100, 'orange': 90}
```

```python
fruit_price = {
    "apple": 100, 
    "banana": 50, 
    "orange": 80
}
del fruit_price["banana"]
print(fruit_price)  # 輸出: {'apple': 100, '
```

```python
# 建立一個 value 含有 dict 的 dict
user_info = {
    "name": "Harry",
    "age": 20,
    "info": {
        "nickname": "哈利",
        "role": "學生",
        "phones": ["0939595049","0912345678"]
    }
}
print(user_info)  # 輸出: {'name': 'Harry', 'age': 20, 'info': {'nickname': '哈利', 'role': '學生', 'phones': ['0939595049']}}
```

```python
# 取得 nickname 的 value
print(user_info["info"])
print(user_info["info"]["nickname"])  # 輸出: 哈利

```

```python
# 取得所有 phone_number
print(user_info["info"]["phones"][0])
print(user_info["info"]["phones"][1])  # 如果只有一個電話號碼，這行會出現 IndexError
user_phone1 = user_info["info"]["phones"][0]
print(user_phone1)  # 輸出: 0939595049
```

```python
'''
改寫 fruit_dict 每個水果需要 price, quantity/amount, color
apple 有 綠色 和 紅色
orange 有 橘色
tomato 有 紅色
'''

fruit_dict = {
    "apple": {
        "price": 30,
        "quantity": 100,
        "color": ["green", "red"]
    },
    "orange": {
        "price": 20,
        "quantity": 80,
        "color": ["orange"]
    },
    "tomato": {
        "price": 10,
        "quantity": 50,
        "color": ["red"]
        }
}
print(fruit_dict)  # 輸出: {'apple': {'price': 30, 'quantity': 100, 'color': ['green', 'red']}, 'orange': {'price': 20, 'quantity': 80, 'color': ['orange']}, 'tomato': {'price': 10, 'quantity': 50, 'color': ['red']}}

data = {}
```

##### dict 的合併

```python
# dict 的合併 dict.update()
data = {
    "banana": {
        "price": 10,
        "quantity": 60,
        "color": ["yellow"]
    }
}
fruit_dict.update(data)
print(fruit_dict)  # 輸出: {'apple': {'price': 30, 'quantity': 100, 'color': ['green', 'red']}, 'orange': {'price': 20, 'quantity': 80, 'color': ['orange']}, 'tomato': {'price': 10, 'quantity': 50, 'color': ['red']}}
```

##### 另外的 print 方式 `pprint` (pretty-print) 

```python
# 從 Module 引入 function
from pprint import pprint
pprint(fruit_dict)  # 注意順序會被打亂，因為 dict 是無序的
print("="*80)
pprint(fruit_dict, sort_dicts=False)  # 注意順序會被打亂，因為 dict 是無序的
```

```python
'''
新增顏色
apple 有新的品種 顏色要有黃色
tomato 有綠色
'''
print(fruit_dict["apple"]["color"])#確定資料型態是 list
fruit_dict["apple"]["color"].append("yellow")
fruit_dict["tomato"]["color"].append("green")

pprint(fruit_dict, sort_dicts=False)  # 注意順序會被打亂，因為 dict 是無序的
```

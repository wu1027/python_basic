```python
# 初始化一個 list
nums = [1, 2, 3,]
print(nums)  # 輸出: [1, 2, 3]
```

```python
# 新增元素在 list 尾端
nums = [1, 2, 3,]
nums.append(5)
print(nums)
```

```python
#常用
nums = [1, 2, 3, 4, 5, 6]
result = [] 
for num in nums:
    if num % 2 == 0:
        result.append(num)

print(result)  # 輸出: [2, 4, 6]
    
```

```python
# 修改索引位置的元素
nums = [1, 2, 3]
print(nums)  # 輸出: [1, 2, 3]
print(nums[2])#用index找list的值
nums[2] = 100 #指定 index 改值
print(nums)
```

```python
#刪除索引位置的元素 list.pop(<index>)，若不指定 index，則刪除最後一個元素
nums = [1, 2, 3,]
print(nums)

nums.pop(1)  # 刪除索引位置 1 的元素
print(nums)  # 輸出: [1, 3]
```

```python
#list.insert(<index>, <value>)，在指定索引位置插入元素
#在指定索引 1 新增元素 9 其餘元素往後移
nums = [1, 2, 3]
print(nums)  # 輸出: [1, 2, 3]
nums.insert(1, 9)  # 在索引 1 處插入元素 9
print(nums)  # 輸出: [1, 9, 2, 3]
```

```python
# 移除指定的值 list.remove(<value>)
nums = [1, 2, 3,]
print(nums)
nums.remove(2)  # 移除值為 2 的元素
print(nums)  # 輸出: [1, 3]
```

```python
nums = [1, 2, 3,]
for i in range(10):
    nums.insert(0, 9)  # 在索引 0 處插入元素 9
print(nums)  # 輸出: [9, 9, 9, 9, 9, 9, 9, 9, 9, 9, 1, 2, 3]

nums.remove(9)  # 移除值為 9 的第一個元素
print(nums)  # 輸出: [9, 9, 9, 9, 9, 9, 9, 9, 1, 2, 3]
```

```python
data = [9, 9, 9, 9, 9, 9, 9, 9, 9, 9, 1, 2, 3]
while 9 in data:
    data.remove(9)
print(data) 
```

```python
# 計算 list 內的某個值 共有幾個
data = [9, 9, 9, 9, 9, 9, 9, 9, 9, 9, 1, 2, 3]
print(data.count(9))  # 輸出: 10

#移除多餘的 9 只留一個
while data.count(9) > 1:
    data.remove(9)
print(data)  # 輸出: [9, 1, 2, 3]

```

```python
data = [9, 9, 9, 9, 9, 9, 9, 9, 9, 9, 1, 2, 3]
for _ in range(data.count(9) - 1):
    data.remove(9)
print(data)  # 輸出: [9, 1, 2, 3]
```

### list 搭配 for 迴圈

```python
# 建立一個 1~10 的 list
NUMS = []
for i in range(1, 11):
    NUMS.append(i)
print(NUMS)  # 輸出: [1, 2, 3
```

```python
# enumerate() 同時取得 index 與 value
data = [1, 20, 3, 4, 50, 6, 7, 80, 9, 100]

for index, value in enumerate(data):
    if value == 50:
        print(f"{value}在 index {index} 的位置")  # 輸出: 50在 index 4 的位置
        #print(f"index: {index}, value: {value}")  # 輸出: Index: 0, Value: 1
```

```python
lines = ['a', 'b', 'c', 'd', 'e']
for index, value in enumerate(lines):
    if value == 'd':
        print(f"{value}在第 {index + 1} 的位置")  # 輸出: d在第 4 的位置
```

```python
lines = ['a', 'b', 'c', 'd', 'e']
for index, value in enumerate(lines, start = 1):  # 從索引 1 開始
    if value == 'd':
        print(f"{value}在第 {index} 的位置")
```

### List comprehension 列表推導式
`new_list = [expression for item in iterable [if condition]]`  
  
`list的名稱 = [表達式 for 元素 in 迭代對象 [if 判斷式(...的條件)]]`

```python
result = []
for i in range(1, 101):
    result.append(i*2)
print(result)
```

```python
result = [i for i in range(1, 101)]
print(result)
```

```python
result = [i for i in range(1, 101) if i % 2 == 0]
print(result)
```

```python
# 多組文字
word1 = 'Hello'
word2 = 'World'

msg = '%s,%s!' % (word1,word2)
print(msg)
```

```python
# 整數
price = 50

msg = "蘋果 %d 元" % (price) # 放字串會導致錯誤
print(msg)
```

```python
# 文字與整數
name = "Harry"
age = 26

result = "%s is %d years old." % (name, age)
print(result)
```

```python
# 指定寬度 (維持 10 個字元長度，預設向右看齊)
msg = "[%10s]" % ("Hello")
print(msg)
```

```python
# 靠左對齊 (維持 10 個字元長度，向左對齊)

```

```python
# 指定浮點數位數
result = "%.2f" % (3.14256)
print(result)
```

```python
# 指定文字長度上限 (只有文字，才在格式化字串中加入「.」來限定字串長度)
result = "%.3s" % ("Hello")
print(result)
```

```python
# 空白補 0
result = "[%06.2f]" % (3.14256)

print(result)
```

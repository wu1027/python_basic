```python
# 建立 function 3 #帶參數
def hello_user(name):
    print(f"Hello, {name}")

    hello_user("harry")  # 呼叫 function 3

```

```python
# 建立 function 4
def say_hello(name, greeting='good morning'): # 要預設參數值的參數 一定要放在後面
    print(f"{greeting}, {name}")

#say_hello() # 沒給參數會報錯
say_hello("harry")  # 呼叫 function 4
say_hello("harry", "good evening")  # 呼叫 function 4，並傳入自訂的 greeting
```

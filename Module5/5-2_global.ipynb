```python
# 全域變數
name = "harry"

# 定義函式
def say_hello()
    name = "harry" #區域變數 不會影響到全域變數

    print(f"hello{name}") # 因為區域變數內沒有
# 執行函式
say_hello()

# 輸出 name
print(name)
```

##### 使用 global 讓函式存取全域變數的值

```python
# 全域變數
name = "harry"

# 定義函式
def say_hello():
    global name
    name = "alex"

# 執行函式
say_hello()

# 輸出 name

```

邏輯運算子 and / or / not

```python
role = "vip"
age = 15

# 只要其中一個有符合，就執行區塊內程式碼
if role == "vip" or age > 18:
    print("or:符合資格")

if role == "vip" and age > 18:
    print("And:符合資格") # and 兩個條件都符合
```

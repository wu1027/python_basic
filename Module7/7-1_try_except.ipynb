##### try...except...

```python
# 可以連續定義數個例外處理機制
# 最符合的例外事件類別會先拋出
try:
    x = 10
    print(y)
    print(x) #上面有錯 不執行該行
except: # 捕捉所有的錯誤

    print("發生錯誤")
```

```python
try:
    x = 10
    print(x[0])
    print(x)
    print(y)
#except Exception as e: #  錯誤用法
    # print("發生錯誤")
except NameError: #捕捉特定錯誤類別
    print("發生 namerror")
except TypeError:
    print("發生 typeError")
except Exception as e: # 一律放最下方
    print("發生錯誤")
```

```python
try:
    x = 10
    print(x[0])
    print(x)
    print(y)
    # as   把 exception 產生的資料賦值給 e
except Exception as e: #捕捉大部分錯誤類別
    print(f"發生錯誤: {e}")

```

```python
# 多層例外處理:
try:
    x = 10
    try: # 不影響原本程式 單獨使用
        print(y)
    except NameError as e:
        print(e)
    print(x)
    
```

| 用法 | 說明 |
|:---:|:---:|
| `range(m)` | 走訪 0 到 m-1 的數字 |
| `range(n, m)` | 走訪 n 到 m-1 的數字  |
| `range(n, m, step)` | 以step為間距，走訪 n 到 m-1 的數字 |

##### for

```python
# range(n)
for i in range(5):
    print(i, end=", ")

```

```python
# range(n, m)
# for<變數名稱> in <要邊歷的物件>
#1 ~ 10的數字
for n in range(1, 11):
    print(n, end=",")
```

```python
# range(n, m, step)
for i in range(1, 16, 2):
    print(i , end=",")
```

```python
for i  in range(5, 10, 3):
    print(5 , end=",")
```

### 小試身手 for  
請產出 a 開頭的 n 個偶數(含 a 本身)  
Ex:  
a = 5, n = 3  
Output : 6 8 10

a = 4, n = 5  
Output : 4 6 8 10 12

```python
a = 5
n = 3

for i in range(n):
    if a % 2 == 0:
        #迴圈第一次執行 4 + 0*2
        #迴圈第二次執行 4 + 1*2
        #迴圈第三次執行 4 + 2*2
        #迴圈第四次執行 4 + 3*2
        #迴圈第五次執行 4 + 4*2
      print(a + i*2, end = " ")
    else:
        print(a + i*2 + 1, end = " ")    
```

```python
a = 5
n = 3

if a % 2 != 0:
    a += 1

    for i in range(n):
        print(a + i*2, end=" ")
```

##### while

```python
n = 1 

while n <= 5: 
    print(n,end=" , ")
    n += 1 #結束的條件 讓n遞增
```

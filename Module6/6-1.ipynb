```python
# 類別名稱
class car:
    # 類別屬性 (Class Attribute)
    wheels = 4

    # __magic_method
    def __init__(self, brand, model, price):  # 建立物件/某個類別的實例 預設第一時間要執行的函式
        # 實例屬性
        self.brand = brand
        self.model = model
        self.price = price

    # 方法 (Method)
    def display_info(self):
        print(f"{self.brand}-{self.model} has {self.wheels} wheels.")

    def show_discount_price(self, discount):
        print(f"{self.brand}-{self.model} price is ${int(discount * self.price):,}")


print(car.wheels)


car1 = car("toyota", "camry", 700000)  # 建立物件 (car class的一個實例)
car2 = car("benz", "c300", 2700000)

print(car1.wheels)
print(car1.brand)
print(car1.model)

print(car2.wheels)
print(car2.brand)
print(car2.model)

car1.display_info()
car2.display_info()

car1.show_discount_price(0.8)
car1.show_discount_price(1.1)
```
